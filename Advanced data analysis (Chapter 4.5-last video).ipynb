### Discrete and continous data?

Discrete data: countable/can be segmented in to disstinct values
Continous data: measurable, may have untis and cannot be segmented

Continuous data cannot be naturally segmented because it flows smoothly without gaps. 
To segment it, you have to artificially force it into ranges or bins (like grouping ages into "20–29" and "30–39")

### Aggregate function

### you need to get sum, mean, min and maximum values for insurance charges grouped by region and sex?

```python
insurance_data.groupby(["region","sex"])["charges"].aggregate(func=["count","mean","max","min","sum"]).round(2)
```

```python
insurance_data.groupby(["region","sex"])["charges"].agg(func=["count","mean","max","min","sum"]).round(2)
```

### you need to get sum, mean, min and maximum values for insurance charges grouped by region and sex? You need to give the name for the functions that is called not the default name?

```python
insurance_data.groupby(["region","sex"])["charges"].agg(func=[("Count","count"),("Average","mean"),("Maximum value","max"),("Minimum value","min"),("Total","sum")]).round(2)
```

### What is the main use of crosstab how it differs from pivot and groupby?

Groupby and pivot table need a continous variable to perform a aggregagate operation and provide result. 
It can have many categories but it needs a continous data to give proper result

Crosstab can compare the two categorical features to returns their frequencies.

this is a function of panda and not a function of dataframe/Series

### Can we use aggregate function for the two categorical data?

### How to use apply to apply a column? is apply a panda function or a data structure function? What we have to add in apply functuin?

Apply is a DataFrame function
We need to add user defined function as a argument

```python
def get(x):
        if x>20:
            print("Eliagible")
        else:
            print("Not eligible")
```

```python

insurance_data["age"].apply(get)
```

### How to assign the above to a new column?

```python
insurance_data["Eligibilty criteria"]=insurance_data["age"].apply(get)
```

### Why it is giving the value as none for new column?

### Lambda function

### What is lambda function

1. Anonymous/nameless function (no need to give any name for the function as we give for user defined)
2. Single line expression should be given for the function

### Example for the lambda function for the below line how can we replace the user defined fucntion with the lambda?
insurance_data["Eligibility criteria"]=insurance_data.age.apply(get)

```python
insurance_data["EC"]=insurance_data.age.apply(lambda x:"Eligible" if x>20 else "Not eligible")
insurance_data
```

### How to create a function in x variable where two numbers are given and it compares and says greater or lesser? using lambda function?

```python
x=lambda a,b:"Greater" if a>b else "Lesser"
print(x(4,5))
```
