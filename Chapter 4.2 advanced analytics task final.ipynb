### Merging function example

```python
import pandas as pd
df_employees = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Department': ['HR', 'IT', 'Finance', 'IT']
})

df_salaries = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 105],
    'Salary': [70000, 85000, 65000, 90000]
})
```

### Merge inner join
Only matching rows

```python
df_inner=df_employees.merge(df_salaries,how="inner")
df_inner
```

### Merge outer join
Matching and unmatching rows

```python
df_outer=df_employees.merge(df_salaries,on="Employee_ID",how="outer")
df_outer
```

### Left join
All left rows + matching rows on the right

```python
df_left=df_employees.merge(df_salaries,on="Employee_ID",how="left")
df_left
```

### Right Join
All right rows + matching rows on the left

```python
df_right=pd.DataFrame.merge(df_employees,df_salaries,on="Employee_ID",how="right")
df_right
```

```python
df_employees_02 = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Department': ['HR', 'IT', 'Finance', 'IT']
})

df_salaries_02 = pd.DataFrame({
    'E_ID': [101, 102, 103, 105],
    'Salary': [70000, 85000, 65000, 90000]
})

df_names_02 = pd.DataFrame({
    'Employee_ID': [101, 106, 103, 108],
    'Name': ['Alice', 'king', 'Charlie', 'Russal'],
    'Department': ['HR', 'IT', 'Finance', 'IT']
})
```

### Using left on and right on to do outer join

```python
df_merged_outer=pd.merge(df_employees_02,df_salaries_02,left_on="Employee_ID",right_on="E_ID",how="outer")
df_merged_outer["Employee_ID"]=df_merged_outer["Employee_ID"].fillna(df_merged_outer["E_ID"]) # trying to comnine Eid with Employee id
df_merged_outer.drop("E_ID",axis=1,inplace=True)
df_merged_outer
```

### Suffix parameter in merge function

```python
df_merged_suffix=pd.merge(df_employees_02,df_names_02,on="Employee_ID",how="outer")
df_merged_suffix
```

```python
df_merged_suffix=pd.merge(df_employees_02,df_names_02,on="Employee_ID",how="outer",suffixes=("First","Last"))
df_merged_suffix
```

### What is lambda function?

A lambda function is a small anonymous function. A lambda function can take any number of arguments, but can only have one expression.

### Simple lambda function example

```python


first_number=2
second_number=3
add=lambda x,y: x+y
add(first_number,second_number)
```

### To do with elements in list

```python


first_list=[2,1]
second_list=[36,6]
add=list(map(lambda x,y: x+y,first_list,second_list)) # use either map, filter or reduce when using list
add
```

### Using lambda in Map function

Use map() when you want to transform every single item in a list into something new.

**Map Example 1 (titanic): Creating a customized column named "description" to include both passenger genders and passenger class**

```python
passenger_pairs=list(zip(Titanic["sex"],Titanic["pclass"]))
Titanic["description"]=list(map(lambda x: f"A {x[0]} passenger traveling in {x[1]} class", passenger_pairs))
Titanic
```

### MAp Example 2 (healthxp): Calculate the difference between Year and current year

```python
Health_information["Year_difference"]=list(map(lambda x: 2026-x, Health_information["Year"]))
Health_information
```

### Using lambda in Filter function

### Filter Example 1 (titanic): Isolating adult male survivors

```python
import pandas as pd
Titanic_alive=pd.DataFrame(filter(lambda x: x.sex=="male" and x.alive=="yes" and x.age>18,Titanic.itertuples(index=False)))
Titanic_alive
```

### Filter Example 2 (healthxp): Filter the country USA with life expectancy greater than 60 years using lambda filter function

```python
Life_expectancy_USA= pd.DataFrame(filter(lambda x: x.Country == "USA" and x.Life_Expectancy > 60.0, Health_information.itertuples(index=False)))
Life_expectancy_USA
```

### Using lambda in Reduce function

```python
from functools import reduce
list1=[1,2,3,4]
add=reduce(lambda x,y:x+y,list1)
add

```

### Reduce Example 1 : Find the Maximum Fare Paid by a Survivor (Titanic Dataset)

```python
survivor_rows = Titanic[Titanic["alive"] == "yes"]
Survivor_max_expense=reduce(lambda x,y: x if x>y.fare else y.fare,survivor_rows.itertuples(index=False),0.0)
Survivor_max_expense
```
