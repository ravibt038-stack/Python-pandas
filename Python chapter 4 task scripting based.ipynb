### 2.	Create a list of numbers and print only even numbers using a for loop

```python
def even_numbers(start=1,stop=50):
    print("Your list of even numbers are")
    for i in range (start,stop):
        if i%2==0:
            print(i)
```

```python
even_numbers(2,20)
```

### 3.	Write a program that handles division by zero using try-except?

```python
def division(numerator=1,denominator=0):
    try:
        Answer=numerator/denominator
        print (Answer)
    except ZeroDivisionError:
        print("sorry 0 is not accepted as denominator, please input correct denominator")
```

```python
division()
```

### 4.	Given a string, count how many vowels it has

```python
def vowels():
    user_string=input("Enter the word to check how many vowels")
    vowels=["a","e","i","o","u"]
    vowel_count=0
    vowel_present=[]
    for string in user_string:
        for letters in vowels:
            if string==letters:
                vowel_count=vowel_count+1
                vowel_present.append(string)
    print(f"You have total {vowel_count} and those vowels are {vowel_present}")
  
        
```

```python
vowels()
```

### 5.	Reverse a list without using built-in functions

```python
def reverse():
    user_string=input("Enter the word to reverse")
    reverse=[]
    for string in range(0,len(user_string)):
        reverse.append(user_string[-string-1])
    reverse="".join(reverse)
    print("Your reverse word is {}".format(reverse))
   
    
   
```

```python
reverse()
```

### 9.	Replace all missing values in a DataFrame with the column’s mean.

```python
import pandas as pd
Employee_salary=pd.read_csv("Employee_sample.csv")
Employee_salary
```

```python
#replacing the string $ and , characters with empty character
```

```python
Employee_salary["Monthly Salary"]=Employee_salary["Monthly Salary"].str.replace(r"[$,]","",regex=True).str.strip().astype("float") 
```

```python
 #mean of salary 
```

```python
Employee_salary["Monthly Salary"].mean()
```

```python
 # Missing salary replaced with mean of salary 
```

```python
Employee_salary["Monthly Salary"].replace(float("nan"),Employee_salary["Monthly Salary"].mean(),inplace=True) #mean of salary was replaced in 
```

```python
 # Missing age values replaced with mean age
```

```python
Employee_salary["Age"].replace(float("nan"),Employee_salary["Age"].mean(),inplace=True)
```
